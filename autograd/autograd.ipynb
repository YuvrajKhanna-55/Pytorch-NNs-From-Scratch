```python
import torch 
x=torch.tensor(3.0,requires_grad=True) #differentiation wrt to x
y=x**2
y.backward()
x.grad
```

```python
import torch # torch uses the computational graph by forward and backward pass
x=torch.tensor(4.0,requires_grad=True)
y=x**2
z=torch.sin(y)
z.backward()
x.grad
```

```python
y.grad # you aren't able to find the differentiation at intermediate level 
```

```python
#vector tensor
import torch # torch uses the computational graph by forward and backward pass
x=torch.tensor([1.0,2.0,3.0],requires_grad=True)
y=(x**2).mean()
y.backward()
x.grad
```

```python
#clearing grad
import torch # torch uses the computational graph by forward and backward pass
x=torch.tensor(3.0,requires_grad=True)
y=x**2
y.backward()
x.grad
x.grad.zero_()# clear the gradient after every forward pass
```
