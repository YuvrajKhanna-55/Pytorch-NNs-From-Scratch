```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from collections import Counter
from torch.utils.data import Dataset, DataLoader
from nltk.tokenize import word_tokenize
import nltk
```

```python
document = """Artificial Intelligence and Reinforcement Learning are among the most important and rapidly growing technologies in modern computer science. Artificial Intelligence, commonly known as AI, refers to the development of machines and software systems that can perform tasks requiring human intelligence such as learning, reasoning, problem-solving, decision-making, speech recognition, and understanding natural language. AI systems are designed to analyze large amounts of data, identify patterns, and make intelligent decisions with minimal human intervention. AI has become an essential part of various industries including healthcare, education, finance, transportation, agriculture, and entertainment. Applications such as virtual assistants, recommendation systems, self-driving cars, chatbots, and facial recognition systems are all examples of AI being used in everyday life. One of the most important branches of AI is Machine Learning, where computers are trained to learn from data instead of being explicitly programmed. Reinforcement Learning (RL) is a specialized area of machine learning in which an intelligent agent learns by interacting with its environment through trial and error. In reinforcement learning, the agent performs actions and receives rewards or penalties depending on the success of those actions. The main goal of the agent is to maximize rewards by learning the best possible strategy over time. This learning process is inspired by human behavior, where individuals learn from experience and improve through feedback. The main elements of reinforcement learning include the agent, environment, state, action, and reward. RL systems continuously balance exploration, which involves trying new actions, and exploitation, which involves using already learned successful actions. Reinforcement learning is widely used in robotics, gaming, autonomous vehicles, industrial automation, and financial systems. AI systems powered by reinforcement learning have achieved remarkable success in games such as chess and Go by defeating world champions through self-learning techniques. The combination of AI and reinforcement learning has significantly advanced the development of intelligent systems capable of adapting to dynamic environments and making autonomous decisions. However, these technologies also face challenges such as high computational costs, ethical concerns, data privacy issues, and the risk of biased decision-making. Despite these challenges, AI and reinforcement learning continue to transform the modern world by improving efficiency, automation, accuracy, and innovation across multiple domains. Their future potential is enormous, and they are expected to play a major role in the advancement of smart technologies and intelligent automation systems in the coming years.

"""
```

```python
# tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
tokens = word_tokenize(document.lower())
```

```python
# vocab
vocab = {'<unk>':0}

for token in Counter(tokens).keys():
  if token not in vocab:
    vocab[token] = len(vocab)

print(vocab)
print(len(vocab))
```

```python
input_sentences = document.split('\n')
input_sentences
```

```python
for sentence in input_sentences:
  print(word_tokenize(sentence.lower()))
```

```python
def text_to_indices(sentence, vocab):

  numerical_sentence = []

  for token in sentence:
    if token in vocab:
      numerical_sentence.append(vocab[token])
    else:
      numerical_sentence.append(vocab['<unk>'])

  return numerical_sentence

input_numerical_sentences = []

for sentence in input_sentences:
  input_numerical_sentences.append(text_to_indices(word_tokenize(sentence.lower()), vocab))

len(input_numerical_sentences)
```

```python
training_sequence=[]
for sentence in input_numerical_sentences:
  for i in range(1,len(sentence)):
    training_sequence.append(sentence[:i+1])

print(training_sequence[:10])
print(len(training_sequence))
```

```python
len_list=[]
for sequence in training_sequence:
  len_list.append(len(sequence))

print(max(len_list))
```

```python
padded_training_sequence = []
for sequence in training_sequence:

  padded_training_sequence.append([0]*(max(len_list) - len(sequence)) + sequence)

len(padded_training_sequence[10])
```

```python
padded_training_sequence = torch.tensor(padded_training_sequence, dtype=torch.long)
padded_training_sequence
```

```python
X = padded_training_sequence[:, :-1]
y = padded_training_sequence[:,-1]
```

```python
class CustomDataset(Dataset):

  def __init__(self, X, y):
    self.X = X
    self.y = y

  def __len__(self):
    return self.X.shape[0]

  def __getitem__(self, idx):
    return self.X[idx], self.y[idx]


dataset = CustomDataset(X,y)

dataloader = DataLoader(dataset, batch_size=32, shuffle=True)
```

```python
class LSTMModel(nn.Module):

  def __init__(self, vocab_size):
    super().__init__()
    self.embedding = nn.Embedding(vocab_size, 100)
    self.lstm = nn.LSTM(100, 150, batch_first=True)
    self.fc = nn.Linear(150, vocab_size)

  def forward(self, x):
    embedded = self.embedding(x)
    intermediate_hidden_states, (final_hidden_state, final_cell_state) = self.lstm(embedded)
    output = self.fc(final_hidden_state.squeeze(0))
    return output
```

```python
model = LSTMModel(len(vocab))
```

```python
epochs = 25
learning_rate = 0.001

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
```

```python
# training loop

for epoch in range(epochs):
  total_loss = 0

  for batch_x, batch_y in dataloader:

    batch_x, batch_y = batch_x, batch_y

    optimizer.zero_grad()

    output = model(batch_x)

    loss = criterion(output, batch_y)

    loss.backward()

    optimizer.step()

    total_loss = total_loss + loss.item()

  print(f"Epoch: {epoch + 1}, Loss: {total_loss:.4f}")
```

```python
# prediction

def prediction(model, vocab, text):

  # tokenize
  tokenized_text = word_tokenize(text.lower())

  # text -> numerical indices
  numerical_text = text_to_indices(tokenized_text, vocab)

  # padding
  padded_text = torch.tensor([0] * (61 - len(numerical_text)) + numerical_text, dtype=torch.long).unsqueeze(0)

  # send to model
  output = model(padded_text)

  # predicted index
  value, index = torch.max(output, dim=1)

  # merge with text
  return text + " " + list(vocab.keys())[index]

prediction(model, vocab, "Artificial Intelligence and Reinforcement Learning")
```

```python
dataloader1 = DataLoader(dataset, batch_size=32, shuffle=False)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
# Function to calculate accuracy
def calculate_accuracy(model, dataloader, device):
    model.eval()  # Set the model to evaluation mode
    correct = 0
    total = 0

    with torch.no_grad():  # No need to compute gradients
        for batch_x, batch_y in dataloader1:
            batch_x, batch_y = batch_x, batch_y

            # Get model predictions
            outputs = model(batch_x)

            # Get the predicted word indices
            _, predicted = torch.max(outputs, dim=1)

            # Compare with actual labels
            correct += (predicted == batch_y).sum().item()
            total += batch_y.size(0)

    accuracy = correct / total * 100
    return accuracy

# Compute accuracy
accuracy = calculate_accuracy(model, dataloader,device )
print(f"Model Accuracy: {accuracy:.2f}%")
```
