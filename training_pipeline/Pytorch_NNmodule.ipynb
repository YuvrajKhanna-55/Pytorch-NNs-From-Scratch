```python
#The torch.nn module in PyTorch provides the essential building blocks for creating and training neural networks.
#It defines the nn.Module base class, which is the foundational element for all network layers and models.
```

```python
#build a nn using nn module single neuron.
import torch
import torch.nn as nn

class Model(nn.Module):
  def __init__(self,num_features):

    super().__init__() #child class
    self.linear=nn.Linear(num_features,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self, features):
    out=self.linear(features)
    out=self.sigmoid(out)
    return out
```

```python
features=torch.rand(10,5)# five inputs
model=Model(features.shape[1])

model(features)
```

```python
print(model.linear.bias)
print(model.linear.weight)
```

```python
## nn with an hidden layer with 18weights and 4 bias
# hidden layer use relu and output activation using sigmoid activation function.
import torch
import torch.nn as nn
class Model2(nn.Module):
  def __init__(self,num_features):

    super().__init__()
    self.linear1=nn.Linear(num_features,3)#layer 1
    self.relu=nn.ReLU()

    self.linear2=nn.Linear(3,1) # layer 2
    self.sigmoid=nn.Sigmoid()

  def forward(self,features):
    out=self.linear1(features)
    out=self.relu(out)
    out=self.linear2(out)
    out=self.sigmoid(out)

    return out

```

```python
features=torch.rand(10,5)

model=Model2(features.shape[1])
model(features)
```

```python
!pip install torchinfo
from torchinfo import summary
summary(model, input_size=(10,5))
```

```python
# use of nn.Sequential in model defining (sequential container)

class Model3(nn.Module):
  def __init__(self,num_features):

    super().__init__()
    self.network=nn.Sequential(
        nn.Linear(num_features,3),
        nn.ReLU(),
        nn.Linear(3,1),
        nn.Sigmoid()
    )

  def forward(self,features):
    out=self.network
    return out

```

```python
features=torch.rand(10,5)

model3=Model3(features.shape[1])
model3(features)
```
